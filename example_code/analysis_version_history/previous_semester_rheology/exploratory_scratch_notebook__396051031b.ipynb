```python
import sys
!{sys.executable} -m pip install pdfplumber openpyxl -q

import pdfplumber
import pandas as pd

all_tables = []

with pdfplumber.open('2490.pdf') as pdf:
    for i, page in enumerate(pdf.pages):
        tables = page.extract_tables()
        for j, table in enumerate(tables):
            df = pd.DataFrame(table)
            df['Page'] = i + 1
            df['Table_on_page'] = j + 1
            all_tables.append(df)
            print(f"Page {i+1}, Table {j+1} — {len(df)} rows, {len(df.columns)} columns")

if all_tables:
    with pd.ExcelWriter('2490_tables.xlsx', engine='openpyxl') as writer:
        for idx, df in enumerate(all_tables):
            sheet_name = f"P{df['Page'].iloc[0]}_T{df['Table_on_page'].iloc[0]}"
            df.to_excel(writer, sheet_name=sheet_name, index=False)
    print(f"\nSaved {len(all_tables)} tables to 2490_tables.xlsx")
else:
    print("No tables found in the PDF.")
```
